```python
import pandas as pd

# Load dataset
df = pd.read_csv("smart_home_energy_consumption_large.csv")


df.head()
```

```python
df.info()
```

```python
df = df.dropna()
df.info()
```

```python
df['datetime'] = pd.to_datetime(df['Date'] + ' ' + df['Time'])
```

```python
df = df.set_index('datetime')
```

```python
df.head()
```

```python
df = df.drop(columns=['Date', 'Time'])
```

```python
df['hour'] = df.index.hour
df['day'] = df.index.day
df['weekday'] = df.index.weekday      # 0 = Monday
df['month'] = df.index.month
```

```python
df[['hour', 'day', 'weekday', 'month']].head()
```

```python
df = df.sort_index()
```

```python
df['lag_1'] = df.groupby('Appliance Type')['Energy Consumption (kWh)'].shift(1)
df['lag_24'] = df.groupby('Appliance Type')['Energy Consumption (kWh)'].shift(24)

df['rolling_mean_24'] = (
    df.groupby('Appliance Type')['Energy Consumption (kWh)']
      .shift(1)
      .rolling(window=24)
      .mean()
)
```

```python
df = df.dropna()
```
