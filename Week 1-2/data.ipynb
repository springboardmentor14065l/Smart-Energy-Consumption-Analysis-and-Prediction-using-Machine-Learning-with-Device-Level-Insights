```python
import pandas as pd

# Correct path (because notebook is inside Week 1-2 folder)
raw_path = "../data/raw data/smart_home_energy_consumption_large.csv"

df = pd.read_csv(raw_path)

# Create timestamp
df["timestamp"] = pd.to_datetime(df["Date"] + " " + df["Time"])

# Keep needed columns
df = df[["Appliance Type", "timestamp", "Energy Consumption (kWh)"]]

# Create hourly dataset
hourly_df = df.groupby(
    ["Appliance Type", pd.Grouper(key="timestamp", freq="H")]
)["Energy Consumption (kWh)"].sum().reset_index()

# Save to processed folder
hourly_df.to_csv("../data/processed/hourly_device_energy.csv", index=False)

print("✅ hourly_device_energy.csv created successfully!")
```
