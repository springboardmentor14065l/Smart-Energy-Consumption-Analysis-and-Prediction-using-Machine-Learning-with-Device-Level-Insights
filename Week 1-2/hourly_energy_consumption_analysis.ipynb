```python
import pandas as pd

# Load dataset from Colab Files
df = pd.read_csv('smart_home_energy_consumption_large.csv')

# Verify loading
df.head()
```

```python
import pandas as pd
import matplotlib.pyplot as plt



# timestamp from Date + Time
df['timestamp'] = pd.to_datetime(df['Date'] + ' ' + df['Time'])

# Sort by timestamp
df = df.sort_values('timestamp')

# timestamp as index
df.set_index('timestamp', inplace=True)

# Calculate hourly energy consumption
hourly_energy = df.resample('H')['Energy Consumption (kWh)'].sum()

# Plot hourly energy consumption
plt.figure(figsize=(12,5))
plt.plot(hourly_energy.index, hourly_energy.values)
plt.xlabel('Time (Hourly)')
plt.ylabel('Energy Consumption (kWh)')
plt.title('Hourly Energy Consumption')
plt.grid(True)
plt.tight_layout()
plt.show()

# Preview result
hourly_energy.head()
```
