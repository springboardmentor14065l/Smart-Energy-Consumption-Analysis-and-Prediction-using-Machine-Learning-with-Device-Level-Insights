```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
df = pd.read_excel("consumption.xlsx")
df.head()
```

```python
df = df.rename(columns={"Outdoor Temperature (掳C)": "Outdoor Temperature (C)"})
```

```python
df["Time"] = pd.to_datetime(df["Time"], format="%H:%M:%S").dt.hour
df.rename(columns={"Time": "Hour"}, inplace=True)
```

```python


df['Date'] = pd.to_datetime(df['Date'])
df['Timestamp'] = df['Date'] + pd.to_timedelta(df['Hour'], unit='h')
df.set_index('Timestamp', inplace=True)
```

```python
df = df.sort_index()
print(df.head())
```

```python
# daily_energy = df.resample('D')['Energy Consumption (kWh)'].sum()
```

```python
season_energy = df.groupby("Season")["Energy Consumption (kWh)"].mean()

plt.figure()
season_energy.plot(kind="bar")
plt.xlabel("Season")
plt.ylabel("Energy Consumption (kWh)")
plt.title("Average Energy Consumption by Season")
plt.show()
```

```python
hourly = df.groupby("Hour")["Energy Consumption (kWh)"].mean()

plt.figure()
hourly.plot()
plt.xlabel("Hour of Day")
plt.ylabel("Energy Consumption (kWh)")
plt.title("Hourly Energy Consumption Pattern")
plt.show()
```

```python
household = df.groupby("Household Size")["Energy Consumption (kWh)"].mean()

plt.figure()
household.plot(kind="bar")
plt.xlabel("Household Size")
plt.ylabel("Energy Consumption (kWh)")
plt.title("Household Size vs Energy Consumption")
plt.show()
```

```python
plt.figure()
df.groupby("Appliance Type")["Energy Consumption (kWh)"].mean().plot(kind="bar")
plt.xlabel("Appliance Type")
plt.ylabel("Average Energy Consumption (kWh)")
plt.title("Average Energy Consumption by Appliance")
plt.show()
```

```python
df["Month"] = df["Date"].dt.month

plt.figure()
df.groupby("Month")["Energy Consumption (kWh)"].mean().plot()
plt.xlabel("Month")
plt.ylabel("Average Energy Consumption (kWh)")
plt.title("Monthly Energy Consumption Trend")
plt.show()
```

```python
import pandas as pd

df_energy = pd.read_excel(
    r"C:\Users\user\internship 1 month\infosys internsip\cleaned_sorted_energy_dataset.xlsx",
    parse_dates=['Timestamp']
)

print(df_energy.shape)
print(df_energy['Appliance Type'].unique())
```

```python
# Feature Engineering: Extracting time-based features
df_energy['hour'] = df_energy['Timestamp'].dt.hour
df_energy['day'] = df_energy['Timestamp'].dt.day
df_energy['weekday'] = df_energy['Timestamp'].dt.weekday
df_energy['month'] = df_energy['Timestamp'].dt.month

# Display the first few rows with the new features
print(df_energy[['Timestamp', 'hour', 'day', 'weekday', 'month']].head())
```

```python
df_energy = df_energy.sort_values(by=['Appliance Type', 'Timestamp'])
df_energy['lag_1'] = df_energy.groupby('Appliance Type')['Energy Consumption (kWh)'].shift(1)
# 3. For each 'Appliance Type', create the 'lag_24' feature
df_energy['lag_24'] = df_energy.groupby('Appliance Type')['Energy Consumption (kWh)'].shift(24)
df_energy['rolling_mean_24'] = df_energy.groupby('Appliance Type')['Energy Consumption (kWh)'] \
                                    .transform(lambda x: x.rolling(window=24, min_periods=1).mean().shift(1))
df_energy.dropna(inplace=True)

# 6. Display the first few rows of the DataFrame, including the newly created features
print(df_energy[['Appliance Type', 'Timestamp', 'Energy Consumption (kWh)', 'lag_1', 'lag_24', 'rolling_mean_24']].head())
```
