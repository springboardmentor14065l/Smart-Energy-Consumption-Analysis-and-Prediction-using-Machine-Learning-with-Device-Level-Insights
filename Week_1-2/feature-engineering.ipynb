```python
import pandas as pd
import numpy as np

data = pd.DataFrame({
    "timestamp": [
        "2024-01-01 08:00", "2024-01-01 08:00",
        "2024-01-01 09:00", "2024-01-01 09:00",
        "2024-01-01 10:00", "2024-01-01 10:00",
        "2024-01-01 11:00", "2024-01-01 11:00"
    ],
    "device": [
        "AC", "Fridge",
        "AC", "Fridge",
        "TV", "Fan",
        "Washing Machine", "Light"
    ],
    "energy": [1.5, 0.3, 1.8, 0.4, 0.6, 0.2, 1.2, 0.1]
})

data["timestamp"] = pd.to_datetime(data["timestamp"])
data["hour"] = data["timestamp"].dt.hour

def day_period(hour):
    if 6 <= hour < 12:
        return "Morning"
    elif 12 <= hour < 17:
        return "Afternoon"
    elif 17 <= hour < 22:
        return "Evening"
    else:
        return "Night"

data["day_period"] = data["hour"].apply(day_period)

high_power_devices = ["AC", "Washing Machine"]
data["is_high_power_device"] = data["device"].apply(
    lambda x: 1 if x in high_power_devices else 0
)

device_usage = data["device"].value_counts()
data["device_usage_count"] = data["device"].map(device_usage)

avg_energy_device = data.groupby("device")["energy"].mean()
data["avg_device_energy"] = data["device"].map(avg_energy_device)

data["is_peak_hour"] = data["hour"].apply(
    lambda x: 1 if 8 <= x <= 11 else 0
)

print("Feature Engineered Dataset:")
print(data)
```
